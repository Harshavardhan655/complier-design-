```python

%%writefile lab1b.l
%{
#include <stdio.h>
#include <stdlib.h>

const char *ones[] = {
    "zero", "one", "two", "three", "four",
    "five", "six", "seven", "eight", "nine"
};

const char *teens[] = {
    "ten", "eleven", "twelve", "thirteen", "fourteen",
    "fifteen", "sixteen", "seventeen", "eighteen", "nineteen"
};

const char *tens[] = {
    "", "", "twenty", "thirty", "forty",
    "fifty", "sixty", "seventy", "eighty", "ninety"
};

void print_number(int n)
{
    if (n < 10 || n > 99) {
        printf("Please enter a two-digit number (10-99).\n");
    } else if (n < 20) {
        printf("%s\n", teens[n - 10]);
    } else {
        printf("%s", tens[n / 10]);
        if (n % 10 != 0)
            printf("-%s", ones[n % 10]);
        printf("\n");
    }
}
%}

%%
[0-9]+ {
    if (yyleng == 2)
        print_number(atoi(yytext));
    else
        printf("Invalid input: %s (enter exactly two digits)\n", yytext);
}
[ \t\n]+ ;
. {
    printf("Invalid character: %s\n", yytext);
}
%%

int main(void)
{
    printf("Enter two-digit numbers:\n");
    yylex();
    return 0;
}

int yywrap(void)
{
    return 1;
}
```

```python

%%writefile lab1c.l
%{
#include <stdio.h>
#include <stdlib.h>

int valid_date(int d, int m, int y)
{
    if (y < 1 || m < 1 || m > 12 || d < 1)
        return 0;

    int days[] = {
        31, 28, 31, 30, 31, 30,
        31, 31, 30, 31, 30, 31
    };

    /* Leap year rule */
    if ((y % 400 == 0) ||
        (y % 4 == 0 && y % 100 != 0))
        days[1] = 29;

    return d <= days[m - 1];
}
%}

%%
[0-9]{2}/[0-9]{2}/[0-9]{4} {
    int d, m, y;
    sscanf(yytext, "%d/%d/%d", &d, &m, &y);

    if (valid_date(d, m, y))
        printf("%s : Valid date\n", yytext);
    else
        printf("%s : Invalid date\n", yytext);
}
[^\n \t]+ {
    printf("%s : Invalid format (use DD/MM/YYYY)\n", yytext);
}
[ \t\n]+ ;
%%

int main(void)
{
    printf("Enter dates in DD/MM/YYYY format:\n");
    yylex();
    return 0;
}

int yywrap(void)
{
    return 1;
}
```
